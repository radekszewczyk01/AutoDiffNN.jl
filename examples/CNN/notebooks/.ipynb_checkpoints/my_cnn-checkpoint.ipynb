```julia
using Pkg
Pkg.instantiate()
Pkg.activate("..")
Pkg.develop(path="C:/Users/ahamed/.julia/dev/myExample")
using Random
using LinearAlgebra
using Printf
using Plots
using myExample
using JLD2, Statistics
const MF = myExample.MiniFlux
const AD =  myExample.AutoDiff

println("Loading dataset...")
X_train = load("../data/imdb_dataset_prepared.jld2", "X_train")
y_train = load("../data/imdb_dataset_prepared.jld2", "y_train")
X_test = load("../data/imdb_dataset_prepared.jld2", "X_test")
y_test = load("../data/imdb_dataset_prepared.jld2", "y_test")
embeddings = load("../data/imdb_dataset_prepared.jld2", "embeddings")
vocab = load("../data/imdb_dataset_prepared.jld2", "vocab")

train_data = [(X_train[i], y_train[i]) for i in 1:length(y_train)]
test_data = [(X_test[i], y_test[i]) for i in 1:length(y_test)]
println("Dataset loaded with $(length(train_data)) training samples")
```

```julia


embedding_dim = size(embeddings,1)

# Define reshape and permute operations
function reshape_op(x)
    return AD.BroadcastedOperator(reshape, x, :, 1, 1, name="reshape")
end

function permute_op(x)
    return AD.BroadcastedOperator(permutedims, x, (2,1,3), name="permute")
end

model = MF.Chain(
    MF.Embedding(length(vocab), embedding_dim),
    reshape_op,
    permute_op,
    MF.Conv((3,), embedding_dim, 8, AD.relu),
    MF.MaxPool((8,)),
    MF.Flatten(),
    MF.Dense(128, 1, AD.σ)
)

# Add Glove embeddings to Embedding layer
model.layers[1].weight .= embeddings


# Add Glove embeddings to Embedding layer
model.layers[1].weight .= embeddings

# Define loss function
function binary_cross_entropy(y_true, y_pred)
    return mean(-y_true .* log.(y_pred .+ 1e-10) .- (1 .- y_true) .* log.(1 .- y_pred .+ 1e-10))
end

# Define accuracy function
function accuracy(y_true, y_pred)
    return mean((y_pred .> 0.5) .== (y_true .> 0.5))
end
# Training loop
function train_model(model, X_train, y_train, X_test, y_test; epochs=5, batchsize=64, lr=0.001)
    for epoch in 1:epochs
        total_loss = 0.0
        total_acc = 0.0
        num_samples = 0
        
        # Create batches
        for i in 1:batchsize:length(X_train)
            batch_end = min(i + batchsize - 1, length(X_train))
            x_batch = X_train[i:batch_end]
            y_batch = y_train[i:batch_end]
            
            # Process each sample in the batch
            for j in 1:length(x_batch)
                x_var = AD.Variable(x_batch[j], name="x")

                println("we got here 1")
                # Forward pass
                pred = model(x_var)
                println("we got here 2")
                # Calculate loss and accuracy
                sample_loss = binary_cross_entropy(y_batch[j], pred)
                sample_acc = accuracy(y_batch[j], pred)
                
                # Backward pass and update
                grads = MF.gradient(model) do m
                    binary_cross_entropy(y_batch[j], m(x_var))
                end
                
                # Update parameters
                for (param, grad) in zip(MF.get_parameters(model), grads)
                    param.output .-= lr .* grad
                end
                
                total_loss += sample_loss
                total_acc += sample_acc
            end
            
            num_samples += 1
        end
        
        # Calculate average metrics
        train_loss = total_loss / num_samples
        train_acc = total_acc / num_samples
        
        # Evaluate on test set
        test_loss = 0.0
        test_acc = 0.0
        
        for i in 1:length(X_test)
            x_test_var = AD.Variable(X_test[i], name="x_test")
            test_pred = model(x_test_var)
            test_loss += binary_cross_entropy(y_test[i], test_pred)
            test_acc += accuracy(y_test[i], test_pred)
        end
        
        test_loss /= length(X_test)
        test_acc /= length(X_test)
        
        println(@sprintf("Epoch: %d", epoch))
        println(@sprintf("Train - Loss: %.4f, Accuracy: %.4f", train_loss, train_acc))
        println(@sprintf("Test  - Loss: %.4f, Accuracy: %.4f", test_loss, test_acc))
    end
end
```

```julia

println("Starting training...")
train_model(model, X_train, y_train, X_test, y_test; 
    epochs=5,      # Number of training epochs
    batchsize=64,  # Batch size for training
    lr=0.001       # Learning rate
)

# Optional: Save the trained model
using JLD2
@save "trained_cnn_model.jld2" model

# Optional: Make predictions on new data
function predict(model, text)
    # Convert text to indices using vocab
    indices = [get(vocab, word, 1) for word in split(text)]
    x = AD.Variable(indices, name="x")
    pred = model(x)
    return pred.output[1]  # Return the probability
end

# Example prediction
test_text = "this movie was really great and i enjoyed it very much"
prediction = predict(model, test_text)
println("Prediction for test text: ", prediction)
println("Sentiment: ", prediction > 0.5 ? "Positive" : "Negative")
```
